```python
class Student:
    def study(self):
        print("Student is studying")

s1 = Student()
s1.study()
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Ravi", 20)
print(s1.name)
print(s1.age)
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        self.balance -= amount

acc1 = BankAccount("Siva", 5000)
acc1.deposit(1000)
print(acc1.balance)
```

```python
class Laptop:
    def __init__(self, brand="Dell"):
        self.brand = brand

l1 = Laptop()
print(l1.brand)

l2 = Laptop("HP")
print(l2.brand)
```

```python
class Employee:
    company = "TCS"   # class variable

    def __init__(self, name):
        self.name = name

e1 = Employee("Ram")
e2 = Employee("Sita")

print(e1.company)
print(e2.company)
```

```python
class MathUtils:

    @staticmethod
    def add(a, b):
        return a + b

print(MathUtils.add(5, 3))
```
