```python
r=5
for i in range(1,r+1):
    for j in range (1,i+1):
        if j==1 or i==r or i==j:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```

```python
# # IMMUTABLE TYPES
# #
# # Cannot change directly:
# #
# # int
# # float
# # string
# # tuple
# # MUTABLE TYPES
# #
# # Can change:
# #
# # list
# # dict
# # set

# Mutable vs. Immutable Types
# In Python, mutable objects can be changed "in place," while immutable objects cannot be modified after they are created; any attempt to change them results in the creation of a new object.
#
# GeeksforGeeks
#  +1
# Mutable Types
# These can be modified without changing their unique identity (memory address).
#
# Real Python
#  +1
# List: You can append, remove, or change items.
# Dictionary (dict): You can update values or add/remove key-value pairs.
# Set: You can add or remove unique items.
# Byte Array: A mutable version of the bytes type.
# User-Defined Classes: Generally mutable by default unless specifically designed otherwise.
#
# GeeksforGeeks
#  +4
# Immutable Types
# Once defined, these cannot be altered. If you "change" a string or integer, Python actually creates a brand-new object in memory.
#
# Real Python
#  +1
# Numbers: Includes int, float, complex, and decimal.
# String (str): You cannot change a specific character within a string.
# Tuple: An ordered collection that cannot be modified after creation.
# Frozenset: An immutable version of a set that is hashable (can be used as a dictionary key).
# Boolean: True and False values.
# Bytes: Fixed-length immutable sequences of bytes.
```
